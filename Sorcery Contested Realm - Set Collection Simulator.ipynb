```python
import pandas as pd
import numpy as np
```

```python
#import example cards
cards_df = pd.read_excel("C:\\Users\\janat\\OneDrive\\Desktop\\Pythonics\\cards test.xlsx")
#add Owned column to track the number of each card that you've pulled from a pack
cards_df['Owned'] = [0 for i in range(len(cards_df))]
```

```python
def generate_collection():
    collection = cards_df.copy()
    return collection
```

```python
def generate_pack_rarities():
    #define contents of pack by rarity except last card - 6 commons, 3 exceptionals, and 1 elite or unique
    pack_rarities = ["Common", "Common", "Common", "Common", "Common", "Common", "Exceptional", "Exceptional", "Exceptional"]
    #determine whether the last card in the pack is elite or unique at a 80:20 ratio
    last_card_rarity_names = ["Elite", "Unique"]
    last_card_rarity_freqs = [0.8, 0.2]
    #update the contents of the pack to include the rarity of the last card
    pack_rarities.append(np.random.choice(last_card_rarity_names, p=last_card_rarity_freqs))
    return pack_rarities
```

```python
def generate_pack():
    #determine whether the tenth card is elite or unique
    pack_rarities = generate_pack_rarities()
    #create an empty pack that will be filled with names of cards based on a list of rarities
    pack = []
    #for each rarity in the list of pack rarities, assign a card name (assumes all card names are equally likely within a rarity)
    for card_rarity in pack_rarities:
        df = cards_df[cards_df["Rarity"] == card_rarity].reset_index(drop=True)
        pack.append(df["Name"].iloc[np.random.randint(len(df))])
    return pack
```

```python
def open_pack(collection):
    #determine the contents of the pack to be opened
    pack = generate_pack()
    #for each name in the pack, increment the number of cards associated with that name in the collection of interest
    for card in pack:
        collection.loc[collection["Name"] == card, "Owned"] += 1
```

```python
def evaluate_collection(collection):
    #check to see if there are any cards where Owned == 0
    num_missing_cards = len(collection[collection["Owned"] == 0])
    collection_incomplete = num_missing_cards > 0
    
    return [collection_incomplete]
```

```python
def summarize_collection(collection):
    total_cards = collection["Owned"].sum()
    total_packs = total_cards / 10
    lost_wealth = "$" + str(total_packs * 6)
    print(collection)

    return [total_cards, total_packs, lost_wealth]
```

```python
def collection_simulator():
    x = generate_collection()

    while evaluate_collection(x) == True:
        open_pack(x)

    summarize_collection(x)
```
